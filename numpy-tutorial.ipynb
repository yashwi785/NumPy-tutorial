```python
#arraqy indexing and slicing
```

```python
x=np.array([1,2,3,9,5])
print(x)
```

```python
#indexing
print(x[0])
print(x[4])
print(x[-1]) # negative indexing
```

```python
#slicing: array[start:stop:step]
print(x[1:4:1])
print(x[1:4:2])
print(x[0:4:1])#first three elements
print(x[-3:]) # last 3 element
```

```python
arr2=np.array([[1,2,3],[4,5,6]])
print(arr2)
```

```python
#indexing
print(arr2[0,2])
print(arr2[1,0])
```

```python
#slicing: array

arr2=np.array([[1,2,3],[4,5,6]])
print(arr2)
print() # for space
print(arr2[1:])
print()
print(arr2[1])
print()
print(arr2[:,1])

#2d:(rows,column)
#3d:(height,rows,column)
```

```python
#array stacking and splitting
```

```python
n=np.array([1,2,3,4])
p=np.array([5,6,7,8])

print(np.vstack((n,p))) #vetical stacking
print(np.hstack((n,p)))#horizontal stacking
```

```python
c= np.array([[1,2,3],[6,7,8]])
print(c)
```

```python
print(np.hsplit(c,3))
print(np.vsplit(c,2))
```

```python
split =np.hsplit(c,3)

for s in split:
    print(s)
```
